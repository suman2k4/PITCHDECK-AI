```python
import fitz  # PyMuPDF
from docx import Document
from unstructured.partition.pptx import partition_pptx
import os

def extract_text(file_path: str) -> str:
    if file_path.endswith(".pdf"):
        doc = fitz.open(file_path)
        return "".join([page.get_text() for page in doc])
    elif file_path.endswith(".docx"):
        return "\n".join([p.text for p in Document(file_path).paragraphs])
    elif file_path.endswith(".pptx"):
        return "\n".join([el.text for el in partition_pptx(file_path)])
    else:  # .txt
        with open(file_path, encoding="utf-8") as f:
            return f.read()
```

```python
import os

if __name__ == "__main__":
    # 📥 Replace with your actual input file path
    sample_file = '../input/NVCPitchDeckTemplate.pdf'
    parsed_text = extract_text(sample_file)  # Make sure this function is defined/imported

    # ✅ Define output directory and make sure it exists
    output_dir = "../input"
    os.makedirs(output_dir, exist_ok=True)

    # ✅ Save parsed text to: ../Pitchdrill/output/parsed_text.txt
    output_path = os.path.join(output_dir, "parsed_text.txt")
    with open(output_path, "w", encoding="utf-8") as f:
        f.write(parsed_text)

    print(f"Parsed text saved to: {output_path}")

```

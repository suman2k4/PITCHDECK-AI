```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from typing import List, Optional
import os

class TextChunker:
    """
    A utility class for splitting text into chunks with configurable size and overlap.
    
    Example usage:
    >>> chunker = TextChunker(chunk_size=1000, chunk_overlap=200)
    >>> chunks = chunker.chunk_text(large_text_document)
    """
    
    def __init__(self, chunk_size: int = 1000, chunk_overlap: int = 200):
        """
        Initialize the text splitter with desired chunk parameters.
        
        Args:
            chunk_size: Maximum size of each chunk (in characters)
            chunk_overlap: Overlap between consecutive chunks (in characters)
        """
        self.splitter = RecursiveCharacterTextSplitter(
            chunk_size=chunk_size,
            chunk_overlap=chunk_overlap
        )
    
    def chunk_text(self, text: str) -> List[str]:
        """Split a single text string into chunks."""
        return self.splitter.split_text(text)
    
    def chunk_file(self, file_path: str, encoding: str = "utf-8") -> List[str]:
        """Read and chunk text directly from a file."""
        with open(file_path, 'r', encoding=encoding) as f:
            text = f.read()
        return self.chunk_text(text)
    
    def save_chunks(self, chunks: List[str], output_dir: str, base_name: str = "chunk"):
        """
        Save chunks to individual files with sequential numbering.
        
        Args:
            chunks: List of text chunks to save
            output_dir: Directory to save chunk files
            base_name: Prefix for chunk filenames
        """
        os.makedirs(output_dir, exist_ok=True)
        
        for i, chunk in enumerate(chunks, start=1):
            output_path = os.path.join(output_dir, f"{base_name}_{i:03d}.txt")
            with open(output_path, 'w', encoding='utf-8') as f:
                f.write(chunk)
            print(f"Saved {output_path} ({len(chunk)} characters)")
```

```python
if __name__ == "__main__":
    # Path to your parsed text file
    input_file = '../input/parsed_text.txt'

    # Output directory for saving the chunks
    output_directory = "../input/output_chunks"

    # Initialize the chunker
    chunker = TextChunker(chunk_size=300, chunk_overlap=50)

    # Read and chunk the file
    chunks = chunker.chunk_file(input_file)

    # Save chunks to separate files
    chunker.save_chunks(chunks, output_dir=output_directory)

    print(f"\nGenerated {len(chunks)} chunks with average length: "
          f"{sum(len(c) for c in chunks)//len(chunks)} characters")
```

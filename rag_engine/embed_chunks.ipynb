```python
from langchain.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
import google.generativeai as genai
from dotenv import load_dotenv
import os

# Load .env
load_dotenv()
genai.configure(api_key=os.getenv("GEMINI_API_KEY"))

# Load text chunks from folder
def load_chunks_from_folder(folder_path: str) -> list[str]:
    chunks = []
    for filename in sorted(os.listdir(folder_path)):
        if filename.endswith(".txt"):
            with open(os.path.join(folder_path, filename), 'r', encoding='utf-8') as f:
                chunks.append(f.read())
    return chunks

# Create vector store
def create_vector_store(chunks: list[str]):
    embeddings = GoogleGenerativeAIEmbeddings(model="models/embedding-001",google_api_key=os.getenv("GEMINI_API_KEY"))
    return FAISS.from_texts(chunks, embeddings)

if __name__ == "__main__":
    chunks = load_chunks_from_folder("../input/output_chunks")
    vector_store = create_vector_store(chunks)
    vector_store.save_local("../input/faiss_index")

    print(f"\nStored {len(chunks)} chunks in FAISS.")
```
